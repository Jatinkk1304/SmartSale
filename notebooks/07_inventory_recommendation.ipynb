### # Step 9 — Inventory Recommendation

In this step, we will use predicted sales demand to calculate recommended additional stock.

The recommendation will use:
- Predicted Sales
- Safety Stock
- Current Stock

Formula:

Recommended Additional Stock = Predicted Demand + Safety Stock - Current Stock

Current Stock is treated as a user/business input because historical inventory stock is not available in the dataset.

```python
# Step 9.2: Load Prediction Data

import pandas as pd
import numpy as np

prediction_data = pd.read_csv(
    "../data/processed/sales_predictions.csv"
)

print("Shape:", prediction_data.shape)
print(prediction_data.head())
```

```python
# Step 9.3: Calculate Safety Stock

prediction_data["Safety_Stock"] = (
    prediction_data["Predicted_Sales"] * 0.10
)

print(
    prediction_data[
        ["Store", "Date", "Predicted_Sales", "Safety_Stock"]
    ].head(10)
)
```

```python
# Step 9.4: Enter Current Stock

current_stock = float(
    input("Enter Current Stock: ")
)

print("Current Stock:", current_stock)
```

```python
# Step 9.5: Calculate Recommended Stock

prediction_data["Recommended_Stock"] = (
    prediction_data["Predicted_Sales"]
    + prediction_data["Safety_Stock"]
    - current_stock
)

prediction_data["Recommended_Stock"] = (
    prediction_data["Recommended_Stock"].clip(lower=0)
)

print(
    prediction_data[
        [
            "Store",
            "Date",
            "Predicted_Sales",
            "Safety_Stock",
            "Recommended_Stock"
        ]
    ].head(10)
)
```

```python
# Step 9.6: Create Inventory Summary

inventory_summary = prediction_data[
    [
        "Store",
        "Date",
        "Predicted_Sales",
        "Safety_Stock",
        "Recommended_Stock"
    ]
].copy()

print("Inventory Summary Shape:", inventory_summary.shape)

print(inventory_summary.head(10))
```

```python
# Step 9.7: Save Inventory Recommendations

inventory_summary.to_csv(
    "../data/processed/inventory_recommendations.csv",
    index=False
)

print("Inventory recommendations saved successfully.")
```

```python
# Step 9.8: Final Verification

print("===== STEP 9 FINAL VERIFICATION =====")

print("Inventory Shape:", inventory_summary.shape)

print("Columns:")
print(inventory_summary.columns.tolist())

print("Missing Values:", inventory_summary.isna().sum().sum())

print(
    "Minimum Recommended Stock:",
    inventory_summary["Recommended_Stock"].min()
)

print(
    "Maximum Recommended Stock:",
    inventory_summary["Recommended_Stock"].max()
)

print("Inventory File Saved Successfully.")

print("===== STEP 9 VERIFICATION COMPLETE =====")
```
